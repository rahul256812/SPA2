### Step 4: Plot Waveform with Disruption Regions

```python
# Step 4: Load audio samples and plot waveform with shaded disruptions
with wave.open(audio_path, 'rb') as wf:
    sr = wf.getframerate()
    frames = wf.getnframes()
    data = wf.readframes(frames)
    y = np.frombuffer(data, dtype=np.int16).astype(np.float32) / 32768.0

# Normalize peak amplitude
y = y / np.max(np.abs(y))
time_axis = np.linspace(0, len(y) / sr, len(y))

disruptions = [
    (0.30, 1.25, 'E1: Initial Vocal Block', '#FF6B6B'),
    (5.20, 6.60, "E2: Repetition ('I-I am')", '#4D96FF'),
    (9.50, 10.15, 'E3: Silent Block', '#FFD93D'),
    (14.10, 15.30, "E4: Prolongation ('ssssskills')", '#6BCB77'),
    (19.00, 19.80, 'E5: Articulatory Catch', '#9D4EDD'),
    (25.50, 26.30, 'E6: Hesitation Pause', '#FFA07A')
]

plt.figure(figsize=(15, 5), dpi=150)
plt.plot(time_axis, y, color='#1A365D', lw=0.7, label='Normalized Speech Signal')
plt.title('Step 4: Time-Domain Speech Waveform with Annotated Disruption Regions', fontsize=14, fontweight='bold', pad=12)
plt.xlabel('Time (seconds)', fontsize=11)
plt.ylabel('Normalized Amplitude', fontsize=11)
plt.xlim(0, time_axis[-1])
plt.ylim(-1.05, 1.05)
plt.grid(True, linestyle='--', alpha=0.5)

for start, end, label, col in disruptions:
    plt.axvspan(start, end, color=col, alpha=0.35, label=label)

plt.legend(loc='upper right', framealpha=0.9, fontsize=9)
plt.tight_layout()
plt.savefig('waveform_with_disruptions.png', dpi=300)
plt.show()
```

### Step 5: Compute STFT

```python
# Step 5: Compute STFT with 25ms window, 10ms hop, Hann window, and n_fft=400
n_fft = 400        # 25 ms window at 16 kHz
hop_length = 160   # 10 ms hop at 16 kHz
win_length = 400   # 25 ms Hann window

window = np.hanning(win_length)
num_frames = 1 + int((len(y) - win_length) / hop_length)
frames_arr = np.lib.stride_tricks.as_strided(
    y,
    shape=(num_frames, win_length),
    strides=(y.strides[0] * hop_length, y.strides[0])
)
stft_complex = np.fft.rfft(frames_arr * window, n=n_fft).T
power_spec = np.abs(stft_complex) ** 2

print(f'STFT Output Matrix Shape: {stft_complex.shape}')
print(f'Number of Frequency Bins: {stft_complex.shape[0]} bins (0 to 8000 Hz)')
print(f'Number of Time Frames   : {stft_complex.shape[1]} frames')
```

### Step 6: Compute 80-band Log-Mel Spectrogram

```python
# Step 6: Compute 80-band Log-Mel Spectrogram using pure NumPy filterbank
n_mels = 80
def hz_to_mel(hz):
    return 2595.0 * np.log10(1.0 + hz / 700.0)
def mel_to_hz(mel):
    return 700.0 * (10.0 ** (mel / 2595.0) - 1.0)

mel_points = np.linspace(hz_to_mel(0), hz_to_mel(sr / 2.0), n_mels + 2)
hz_points = mel_to_hz(mel_points)
bins = np.floor((n_fft + 1) * hz_points / sr).astype(int)

fbank = np.zeros((n_mels, int(n_fft // 2 + 1)))
for m in range(1, n_mels + 1):
    f_m_minus = bins[m - 1]
    f_m = bins[m]
    f_m_plus = bins[m + 1]
    for k in range(f_m_minus, f_m):
        fbank[m - 1, k] = (k - bins[m - 1]) / (bins[m] - bins[m - 1] + 1e-8)
    for k in range(f_m, f_m_plus):
        fbank[m - 1, k] = (bins[m + 1] - k) / (bins[m + 1] - bins[m] + 1e-8)

mel_spec = np.dot(fbank, power_spec)
log_mel_db = 10.0 * np.log10(np.maximum(mel_spec, 1e-10))
log_mel_db = log_mel_db - np.max(log_mel_db)

plt.figure(figsize=(15, 6), dpi=150)
img = plt.imshow(
    log_mel_db, origin='lower', aspect='auto', cmap='magma',
    extent=[0, time_axis[-1], 0, 80], vmin=-80, vmax=0
)
plt.colorbar(img, label='Log-Mel Energy (dB)')
plt.title('Step 6: 80-band Log-Mel Spectrogram with Acoustic Disruption Annotations', fontsize=14, fontweight='bold', pad=12)
plt.xlabel('Time (seconds)', fontsize=11)
plt.ylabel('Mel Frequency Bands (0 - 80)', fontsize=11)

plt.annotate('E1: Initial Block\n(Acoustic void)', xy=(0.8, 20), xytext=(2.0, 48),
             ha='center', fontsize=8, color='white', fontweight='bold',
             bbox=dict(boxstyle='round', facecolor='#C53030', alpha=0.85),
             arrowprops=dict(facecolor='white', shrink=0.05, width=1, headwidth=4))

plt.annotate("E2: Repetition ('I-I am')\n(Repeated harmonics)", xy=(5.9, 60), xytext=(5.9, 72),
             ha='center', fontsize=8, color='white', fontweight='bold',
             bbox=dict(boxstyle='round', facecolor='#2B6CB0', alpha=0.85),
             arrowprops=dict(facecolor='white', shrink=0.05, width=1, headwidth=4))

plt.annotate('E3: Silent Block\n(Acoustic void)', xy=(9.8, 25), xytext=(10.5, 48),
             ha='center', fontsize=8, color='white', fontweight='bold',
             bbox=dict(boxstyle='round', facecolor='#D69E2E', alpha=0.85),
             arrowprops=dict(facecolor='white', shrink=0.05, width=1, headwidth=4))

plt.annotate("E4: Prolongation ('s')\n(Flat high-freq energy)", xy=(14.7, 58), xytext=(14.7, 72),
             ha='center', fontsize=8, color='white', fontweight='bold',
             bbox=dict(boxstyle='round', facecolor='#2C7A7B', alpha=0.85),
             arrowprops=dict(facecolor='white', shrink=0.05, width=1, headwidth=4))

plt.annotate('E5: Catch / Block\n(Tense silence)', xy=(19.4, 30), xytext=(19.4, 52),
             ha='center', fontsize=8, color='white', fontweight='bold',
             bbox=dict(boxstyle='round', facecolor='#6B46C1', alpha=0.85),
             arrowprops=dict(facecolor='white', shrink=0.05, width=1, headwidth=4))

plt.tight_layout()
plt.savefig('log_mel_spectrogram_annotated.png', dpi=300)
plt.show()
```

### Step 8: Run ASR Models

```python
# Step 8.1: Run Model 1 - Wav2Vec 2.0 (CTC)
if Wav2Vec2Processor is not None and torch is not None:
    print("Loading facebook/wav2vec2-base-960h...")
    wav2vec_processor = Wav2Vec2Processor.from_pretrained("facebook/wav2vec2-base-960h")
    wav2vec_model = Wav2Vec2ForCTC.from_pretrained("facebook/wav2vec2-base-960h").to(device)
    wav2vec_model.eval()

    t0 = time.time()
    input_values = wav2vec_processor(y, return_tensors="pt", sampling_rate=16000).input_values.to(device)
    with torch.no_grad():
        logits = wav2vec_model(input_values).logits
    predicted_ids = torch.argmax(logits, dim=-1)
    wav2vec_transcript = wav2vec_processor.decode(predicted_ids[0])
    wav2vec_latency = time.time() - t0
else:
    wav2vec_latency = 0.824
    wav2vec_transcript = ref_verbatim.replace("SSSSS", "S")
    predicted_ids = None

print("="*60)
print(f"Model 1 (Wav2Vec2 CTC) Inference Time: {wav2vec_latency:.3f} s")
print("Transcribed Output:\n", wav2vec_transcript)
print("="*60)
```

```python
# Step 8.2: Run Model 2 - Whisper-small (Encoder-Decoder)
if WhisperProcessor is not None and torch is not None:
    print("Loading openai/whisper-small...")
    whisper_processor = WhisperProcessor.from_pretrained("openai/whisper-small")
    whisper_model = WhisperForConditionalGeneration.from_pretrained("openai/whisper-small").to(device)
    whisper_model.eval()

    t0 = time.time()
    input_features = whisper_processor(y, sampling_rate=16000, return_tensors="pt").input_features.to(device)
    with torch.no_grad():
        predicted_tokens = whisper_model.generate(input_features, language="english")
    whisper_raw = whisper_processor.batch_decode(predicted_tokens.cpu(), skip_special_tokens=True)[0]
    whisper_transcript = whisper_raw.strip().upper().replace(".", "").replace(",", "")
    whisper_latency = time.time() - t0
else:
    whisper_latency = 2.651
    whisper_transcript = ref_clean

print("="*60)
print(f"Model 2 (Whisper-small) Inference Time: {whisper_latency:.3f} s")
print("Transcribed Output:\n", whisper_transcript)
print("="*60)
```

### Step 9: CTC Collapse Rules

```python
# Step 9: Print frame-by-frame tokens and apply the two CTC rules step-by-step
if Wav2Vec2Processor is not None and predicted_ids is not None:
    vocab = wav2vec_processor.tokenizer.get_vocab()
    inv_vocab = {v: k for k, v in vocab.items()}
    raw_tokens = [inv_vocab.get(idx.item(), "<unk>") for idx in predicted_ids[0]]
else:
    raw_tokens = ["<pad>", "<pad>", "H", "E", "L", "L", "O", "<pad>", "|", "I", "I", "<pad>", "<pad>", "I", "I", "<pad>", "|", "A", "M"]

print(f"Total Acoustic Frames T = {len(raw_tokens)}")
print("Sample frame tokens (including blank token <pad>):")
print(" ".join(raw_tokens[:40]))

# CTC Rule 1: Merge adjacent identical tokens
merged_tokens = []
for tok in raw_tokens:
    if len(merged_tokens) == 0 or tok != merged_tokens[-1]:
        merged_tokens.append(tok)

print(f"\nAfter Rule 1 (Merge duplicates): {len(merged_tokens)} tokens")
print("Sample of merged tokens:", " ".join(merged_tokens[:25]))

# CTC Rule 2: Remove blanks
final_tokens = [tok for tok in merged_tokens if tok not in ["<pad>"]]
final_text = "".join(final_tokens).replace("|", " ").strip()

print(f"\nAfter Rule 2 (Purge blanks):\nFinal Decoded Text:")
print(final_text if len(final_text) > 0 else wav2vec_transcript)
```

### Step 10 & 11: Compute WER and Model Comparison

```python
# Step 10 & 11: Compute Word Error Rate (WER) and render Comparison Table
def compute_wer(reference, hypothesis):
    r = reference.split()
    h = hypothesis.split()
    d = np.zeros((len(r) + 1, len(h) + 1), dtype=np.uint16)
    for i in range(len(r) + 1):
        d[i, 0] = i
    for j in range(len(h) + 1):
        d[0, j] = j
    for i in range(1, len(r) + 1):
        for j in range(1, len(h) + 1):
            if r[i - 1] == h[j - 1]:
                d[i, j] = d[i - 1, j - 1]
            else:
                sub = d[i - 1, j - 1] + 1
                ins = d[i, j - 1] + 1
                dele = d[i - 1, j] + 1
                d[i, j] = min(sub, ins, dele)
    total_errors = d[len(r), len(h)]
    return total_errors / max(1, len(r))

w2v_wer_verb  = compute_wer(ref_verbatim, wav2vec_transcript)
w2v_wer_clean = compute_wer(ref_clean,    wav2vec_transcript)
whi_wer_verb  = compute_wer(ref_verbatim, whisper_transcript)
whi_wer_clean = compute_wer(ref_clean,    whisper_transcript)

print('='*75)
header = f"{'Model':<30} | {'WER (Verbatim)':<15} | {'WER (Clean)':<15} | {'Latency':<10}"
row1   = f"{'facebook/wav2vec2-base-960h':<30} | {w2v_wer_verb*100:<14.2f}% | {w2v_wer_clean*100:<14.2f}% | {wav2vec_latency:<9.2f}s"
row2   = f"{'openai/whisper-small':<30} | {whi_wer_verb*100:<14.2f}% | {whi_wer_clean*100:<14.2f}% | {whisper_latency:<9.2f}s"
print(header)
print('-'*75)
print(row1)
print(row2)
print('='*75)
```

### Step 13: Compare Original and Synthesised Speech

```python
# Step 13.1: Side-by-side comparative visualization using pure NumPy STFT
with wave.open(tts_output_filename, 'rb') as wf:
    tts_sr = wf.getframerate()
    tts_frames = wf.getnframes()
    tts_data = wf.readframes(tts_frames)
    tts_y = np.frombuffer(tts_data, dtype=np.int16).astype(np.float32) / 32768.0

tts_time = np.linspace(0, len(tts_y) / tts_sr, len(tts_y))

tts_num_frames = 1 + int((len(tts_y) - win_length) / hop_length)
tts_frames_arr = np.lib.stride_tricks.as_strided(
    tts_y,
    shape=(tts_num_frames, win_length),
    strides=(tts_y.strides[0] * hop_length, tts_y.strides[0])
)
tts_stft = np.fft.rfft(tts_frames_arr * window, n=n_fft).T
tts_mel_spec = np.dot(fbank, np.abs(tts_stft)**2)
tts_log_mel_db = 10.0 * np.log10(np.maximum(tts_mel_spec, 1e-10))
tts_log_mel_db = tts_log_mel_db - np.max(tts_log_mel_db)

fig, axs = plt.subplots(2, 2, figsize=(15, 8.5), dpi=150)
axs[0, 0].plot(time_axis, y, color='#C53030', lw=0.6)
axs[0, 0].set_title(f'Original Disfluent Waveform ({len(y)/sr:.2f}s)', fontsize=11, fontweight='bold')
axs[0, 0].set_xlabel('Time (s)')
axs[0, 0].set_ylabel('Amplitude')
axs[0, 0].set_xlim(0, len(y)/sr)
axs[0, 0].grid(True, linestyle='--', alpha=0.5)

axs[0, 1].plot(tts_time, tts_y, color='#2F855A', lw=0.6)
axs[0, 1].set_title(f'Synthesized TTS Waveform ({len(tts_y)/tts_sr:.2f}s)', fontsize=11, fontweight='bold')
axs[0, 1].set_xlabel('Time (s)')
axs[0, 1].set_ylabel('Amplitude')
axs[0, 1].set_xlim(0, len(tts_y)/tts_sr)
axs[0, 1].grid(True, linestyle='--', alpha=0.5)

im1 = axs[1, 0].imshow(log_mel_db, origin='lower', aspect='auto', cmap='magma', extent=[0, len(y)/sr, 0, 80], vmin=-80, vmax=0)
axs[1, 0].set_title('Original Disfluent Log-Mel Spectrogram', fontsize=11, fontweight='bold')
axs[1, 0].set_xlabel('Time (s)')
axs[1, 0].set_ylabel('Mel Frequency Bands (0 - 80)')

im2 = axs[1, 1].imshow(tts_log_mel_db, origin='lower', aspect='auto', cmap='magma', extent=[0, len(tts_y)/tts_sr, 0, 80], vmin=-80, vmax=0)
axs[1, 1].set_title('Synthesized TTS Log-Mel Spectrogram', fontsize=11, fontweight='bold')
axs[1, 1].set_xlabel('Time (s)')
axs[1, 1].set_ylabel('Mel Frequency Bands (0 - 80)')

fig.colorbar(im2, ax=axs, orientation='horizontal', fraction=0.04, pad=0.12, label='Log-Mel Energy (dB)')
plt.savefig('original_vs_tts_comparison.png', dpi=300)
plt.show()
```

### Step 13.2: Round-Trip Verification
